## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.

- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

### Cleaning Data

#### Fixing ... problem



*   Melakukan penghapusan pada kolom workingday karena sudah diwakilkan oleh kolom weekday
*   Menghapus kolom Instan karena hanya berupa nomor urut sehingga tidak memiliki fungsi apapun
*   Menggunakan tipe data int menjadi category pada kolom 'season', 'mnth', 'holiday', 'weekday', 'weathersit' karena mereka semua memiliki data dengan beberapa varian
* Mengubah tipe data dteday yang tadinya 'object' akan diubah menjadi 'datetime'
* Mengganti nama kolom agar lebih mudah terbaca




```python
# Melakukan penghapusan kolom workingday dan instan pada kedua data frame
df_harian.drop(['workingday', 'instant'], axis=1, inplace=True)
df_Perjam.drop(['workingday', 'instant'], axis=1, inplace=True)
```

```python
#Menggunakan tipe data int menjadi category pada kolom 'season', 'mnth', 'holiday', 'weekday', 'weathersit'
columns = ['season', 'mnth', 'holiday', 'weekday', 'weathersit']
for column in columns:
  df_harian[column] = df_harian[column].astype('category')
  df_Perjam[column] = df_Perjam[column].astype("category")
```

```python
#Mengubah tipe data dteday yang tadinya 'object' akan diubah menjadi 'datetime'
df_harian['dteday'] = pd.to_datetime(df_harian['dteday'])
df_Perjam['dteday'] = pd.to_datetime(df_Perjam['dteday']).astype('datetime64[ns]')
```

```python
#Mengganti nama kolom agar lebih mudah terbaca
df_harian.rename(columns={'dteday': 'date', 'yr': 'year', 'mnth': 'month', 'hum': 'humidity', 'cnt': 'count', 'weathersit' : 'weather_situation'}, inplace=True)
df_Perjam.rename(columns=({'dteday': 'date', 'yr': 'year', 'mnth': 'month', 'hum': 'humidity', 'cnt': 'count', 'weathersit' : 'weather_situation', 'hr' : 'hour'}), inplace=True)
```

```python
# Mengkonversi isi kolom agar mudah dipahami
# konversi season menjadi: 1:Spring, 2:Summer, 3:Fall, 4:Winter
df_harian['season'] = df_harian['season'].map({1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'})
df_Perjam['season'] = df_Perjam['season'].map({1: 'Spring', 2: 'Summer', 3: 'Fall', 4: 'Winter'})
```

## Visualization & Explanatory Analysis

### Pertanyaan 1: Bagaimana perbandingan pola penggunaan pelanggan yang melakukan register dengan casual ?

```python
#Menjumlahkan semua elemen dalam column holiday
casual_count = sum(df_harian['casual'])
#Menjumlahkan semua elemen dalam column weekday
registered_count = sum(df_harian['registered'])
#membuat data untuk pieplot
data = [casual_count, registered_count]
label = ['Casual', 'Registered']
plt.pie(data, labels=label, autopct='%1.1f%%')
plt.title('Perbandingan Pola Penggunaan Pada Registered dengan Casual')
plt.show()
```

pada pie chart diatas, didapatkan data bahwa sebanyak 81.2 % dari 100% pelanggan lebih memilih untuk berlangganan dan sisanya sebanyak 18.8% dari 100% pelanggan memilih untuk tidak berlangganan

### Pertanyaan 2: Bagaimana strategi pemasaran untuk perusahaan memilih musim dengan permintaan tertinggi ?

```python
# membuat subplot dengan 1 baris dan 1 kolom, dengan ukuran (30, 10)
fig, ax = plt.subplots(figsize=(30, 10))
# Membuat Bar plot
sns.barplot(
    y = 'count',
    x = 'season',
    data = df_harian,
    ax = ax
)
# Mengatur judul, label y, label x
ax.set_title('Perbandingan Grafik Antar Musim', loc= 'center', fontsize=30)
ax.set_ylabel(None)
ax.set_xlabel(None)
ax.tick_params(axis='x', labelsize=35)
ax.tick_params(axis='y', labelsize=30)
plt.show()
```

pada Bar plot diatas, dapat disimpulkan bahwa strategi pemasaran yang tepat adalah pada Fall Season / Musim Gugur karena permintaan yang tinggi dibandingkan 3 musim lainnya.

### Pertanyaan 3 : Bagaimana pola penggunaan sepeda berdasarkan perbandingan Jam

```python
#membuat bar chart untuk melihat pola penyewaan sepeda berdasarkan jam
fig, ax = plt.subplots(figsize=(30, 10))

# Menjumlahkan pola pengunaan sepeda berdasrkan jam
jumlah_perjam = df_Perjam.groupby('hour')['count'].sum().reset_index()

# menentukan jumlah penguna terbanyak berdasarkan jam
peak_hour = jumlah_perjam.loc[jumlah_perjam['count'].idxmax(), 'hour']

#melakukan pemilihan warna pada data terbanyak
colors = ["#D3D3D3"] * 24
colors[int(peak_hour)] = "#90CAF9"

sns.barplot(
    y = 'count',
    x = 'hour',
    data = df_Perjam,
    hue = 'hour',
    palette = colors,
    legend = False,
    ax=ax
)

ax.set_title('Pola Penggunaan Sepeda Per Jam', loc='center', fontsize=30)
ax.set_ylabel('Jumlah Penyewaan', fontsize=30)
ax.set_xlabel('Jam', fontsize=30)
ax.tick_params(axis='x', labelsize=20)
ax.tick_params(axis='y', labelsize=20)
plt.show()
```

pada data tersebut dapat disimpulkan bahwa, pola penggunaan sepeda lebih banyak pada jam 17.00 sebanyak 450 pelanggan
